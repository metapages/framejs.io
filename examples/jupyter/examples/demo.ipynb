# framejs in Jupyter

Use any saved frame as a notebook widget. Just paste its URL.

```python
from framejs import Frame
```

## 1. Create from a saved short URL

```python
# An "echo" metaframe (built in the framejs.io editor, saved to a short URL):
# it displays its inputs and passes each input key through as an output.
echo = Frame(
    url="https://framejs.app/j/019f61392e0778f8aba8dd7ffe35a83c"
)
echo
```

## 2. Push inputs from Python

```python
echo.set_inputs({"data": [1, 2, 3], "message": "hello from Python"})
```

## 3. Read outputs

```python
# Outputs arrive asynchronously — re-run this cell after a moment if it shows {}
echo.outputs
```

## 4. React to output changes

A callback runs when the widget sends new outputs — but note **where its output
goes**. The callback fires while the kernel is handling a message from the
browser, not while a cell is running, so a bare `print()` has no cell to print
into: JupyterLab files it under **View → Show Log Console**, and some front-ends
drop it entirely. Render into an `Output` widget to see it in the notebook.

```python
import ipywidgets as widgets

# `log` is displayed by this cell (it is the last expression), and the callback
# appends to it. Use `append_stdout` rather than `print` — see the note above.
log = widgets.Output()
echo.on_outputs_change(lambda change: log.append_stdout(f"Got: {change['new']}\n"))
log
```

## 5. Pipe widgets together

```python
# A "doubler" metaframe (built in the editor, saved to a short URL):
# it doubles every number in the "data" input and emits it as "doubled".
doubler = Frame(
    url="https://framejs.io/j/7099ba440f37b858bf33c5fd09ae04077a92318c7abc7f7669117db540d776c9"
)

# Connect echo's "data" output to doubler's "data" input
echo.pipe_to(doubler, output_key="data", input_key="data")

doubler
```

```python
# Push data through the pipeline
echo.set_inputs({"data": [10, 20, 30]})
```
